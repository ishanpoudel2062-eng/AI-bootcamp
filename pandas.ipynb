# DataFrame
Series is like a column, a DataFrame is the whole table. it contains rown and columns.(2-dimension)

```python


import pandas as pd
dic={
   "name":["ishan","ram","sita"],
   "age":[11,12,23],
   "class":[12,15,13]
}
d=pd.DataFrame(dic)
print(d)
```

```python

# indexing  or labeling according to our choice
import pandas as pd
dic={
   "name":["ishan","ram","sita"],
   "age":[11,12,23],
   "class":[12,15,13]
}
d=pd.DataFrame(dic,index=["1","2","3"])
print(d)
```

For acessing specific data we use loc sytnax.


```python
import pandas as pd
dic={
    "name":["ishan","ram","sita"],
       "age":[11,12,23],
       "class":[12,15,13]
    
}
d=pd.DataFrame(dic,index=[1,2,3])
print(d.loc[1])# only printing the index 1
print("------------------------------")
print(d.loc[[1,3]])# printing the value of index 1 and 3
```

# Cleaning Data


cleaining data means fixing the bad data of datasheet.like wrong format of data, duplicate, empty one etc.

```python
# cleaning empty cells
# removing that empty cell row 
import pandas as pd

df = pd.read_csv('data.csv')

new_df = df.dropna()

print(new_df.head(10).to_string()) #below output terminal we can index 2 row is remove as it contains missing value
```

```python
# replacing null value
import pandas as pd

df = pd.read_csv('data.csv')

df.fillna(132, inplace = True)# inplace=true means tell pandas to modify the original data directly
```

in above it replace all the empty one with assign value and now assigning the value to specific one

```python
import pandas as pd

df = pd.read_csv('data.csv')

df.fillna({"Calories": 130}, inplace=True)
```

#  sorting 

sorting means arranging the value in ascending or desceding order based on their column value

```python
import pandas as pd
df=pd.read_csv("data.csv")
df1=df.sort_values("Pulse",ascending=False).head(5)# arranged in descending order
print(df1)
```

# Indexing

Indexing is one of the most important concepts in Pandas. It is how Pandas identifies, selects, and accesses rows and columns in a DataFrame or Series.

If we understand indexing well, we will be able to filter, modify, and analyze data easily.

There are mainly four ways to index data.

1)Column indexing

2)Row indexing

3)loc[]

4)iloc[]

```python
#1 column indexing
import pandas as pd
df=pd.read_csv("data.csv")
print(df["Pulse"].head(10))# print plne one column
print("--------------------------------------")
print(df[["Pulse","Duration"]].head(10))# print multiple column
```

# Row Indexing

 # Using .loc[] (Label-Based Indexing)

.loc[] selects rows and columns by their labels (index names and column names).

```python
import pandas as pd
df=pd.read_csv("data.csv")
print(df.loc[2]) # get 2nd row
```

```python
import pandas as pd
df=pd.read_csv("data.csv")
print(df.loc[2:7])
```

 # .iloc[] (Position-Based Indexing)

iloc=integer location

This means it accesses data using integer positions (0, 1, 2, 3, ...), not row labels or column names.



```python
import pandas as pd
df=pd.read_csv("data.csv")
print(df.iloc[2:7])
```
